## Serpapi

```python
client = serpapi.Client(api_key='388d93152e880cbd0ad9f1be32a68b1dd5a2d5ed108833c74643f04ac03fe647')

keyword = input('Keyword: ')
file_name = input('Nama File (tanpa .csv): ')
total_halaman = int(input('Jumlah halaman yang ingin diambil (misal 5 untuk ~50 berita): '))

all_news = []

for page in range(total_halaman):
    start_offset = page * 10

    result = client.search({
        'q': keyword,
        'cr': 'countryID',
        'tbm': 'nws',
        'tbs': 'qdr:d',
        'start': start_offset
    })

    news_data = result.get('news_results', [])

    if not news_data:
        print(f"Halaman {page + 1} tidak memiliki data lagi. Proses dihentikan.")
        break

    all_news.extend(news_data)
    print(f"Halaman {page + 1} selesai diambil ({len(news_data)} berita).")

# Transform & Load
df = pd.DataFrame(all_news)

if not df.empty:
    directory = '../data/raw/news'
    os.makedirs(directory, exist_ok=True)
    save_path = os.path.join(directory, f'{file_name}.csv')
    df.to_csv(save_path, index=False)

    print(f"\n✅ Total {len(df)} berita berhasil disimpan ke: {save_path}")
    print(df.head())
else:
    print("\n⚠️ Tidak ada berita yang berhasil didapat.")
```

## Kaggle

```python
total_dataset = int(input('Jumlah dataset: '))
for i in range(total_dataset):
    directory = str(input('Directory Kaggle; '))
    file = str(input('nama File Dataset Kaggle: '))
    folder_target = '../data/raw/kaggle'

    try:
        path = kagglehub.dataset_download(directory)
        print(f'path download: {path}')

        full_file_path = os.path.join(path, file + ".csv")
        df_kaggle = pd.read_csv(full_file_path, encoding='utf-8', low_memory=False)

        full_path_save = os.path.join(folder_target, f'{file}.csv')
        df_kaggle.to_csv(full_path_save, index=False)
    except Exception as e:
        print('Error Handling 01...\n', e)
        continue
```
